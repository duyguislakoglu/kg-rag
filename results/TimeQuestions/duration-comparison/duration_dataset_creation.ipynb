```python
from collections import Counter

counts = Counter(tuple(item[:3]) for item in lines)
```

```python
filtered_data = [
    item for item in lines
    if counts[tuple(item[:3])] == 1
]
```

```python
events_with_durations = dict()
to_be_removed = []
all_events = []
for event in filtered_data:
    
    all_events.append(' '.join(event[:3]))
    
    if int(event[3]) == 1 or int(event[4]) == 2916:
        continue
    if event[3] == event[4]:
        continue
    if int(event[4])-int(event[3]) <0:
        continue
    if ' '.join(event[:3]) in events_with_durations.keys():
        print(event)
        to_be_removed.append(' '.join(event[:3]))
        continue
    events_with_durations[' '.join(event[:3])] = (event[3:], int(event[4]) - int(event[3]))
```

```python
len(events_with_durations)

len(to_be_removed)

len(set(to_be_removed))


for key in to_be_removed:
    events_with_durations.pop(key, None)   # None avoids KeyError if key is missing

len(events_with_durations)
```

```python
events_with_durations
```

```python
import random
random.sample(list(events_with_durations.keys()), 2)

longer_templates = [
    "Which event lasted longer, [event1] or [event2]?",
    "Which event took longer, [event1] or [event2]?",
    "Which event had a longer duration, [event1] or [event2]?",
    "Between [event1] and [event2], which lasted longer?",
    "Which event was longer, [event1] or [event2]?",
    "Of [event1] and [event2], which one took longer?"
]

shorter_templates = [
    "Which event lasted for a shorter time, [event1] or [event2]?",
    "Which event took less time, [event1] or [event2]?",
    "Which event had a shorter duration, [event1] or [event2]?",
    "Between [event1] and [event2], which lasted for a shorter time?",
    "Which event was shorter, [event1] or [event2]?",
    "Of [event1] and [event2], which one was shorter?"
]

questions = []
longer_count = 0 
shorter_count = 0 
while True:
    random_keys = random.sample(list(events_with_durations.keys()), 2)
    dur1= events_with_durations[random_keys[0]][1]
    dur2= events_with_durations[random_keys[1]][1]
    rel = ""
    if abs(dur1 - dur2) < 5:
        continue
    if dur1 > dur2:
        if longer_count >= 700:
            continue
        rel = "longer"
        longer_count +=1
        random_template = random.sample(list(longer_templates), 1)[0]

    else:
        rel = "shorter"
        if shorter_count >= 700:
            continue
        shorter_count +=1
        random_template = random.sample(list(shorter_templates), 1)[0]

    question = random_template.replace("event1", random_keys[0]).replace("event2",random_keys[1])
    questions.append(((question,random_keys[0]),random_keys[0]+ str(events_with_durations[random_keys[0]]), random_keys[1]+str(events_with_durations[random_keys[1]])))
    
    question = random_template.replace("event1", random_keys[1]).replace("event2",random_keys[0])
    questions.append(((question,random_keys[0]), random_keys[0] + str(events_with_durations[random_keys[0]]), random_keys[1] + str(events_with_durations[random_keys[1]])))
    
    events_with_durations.pop(random_keys[0], None) 
    events_with_durations.pop(random_keys[1], None) 
    if longer_count >= 700 and shorter_count >= 700:
        break
    
```

```python
lines = []

for question in questions[:2000]:
    
    question_entities = [{
        "name": None,
        "mention": None,
        "span": None, 
        "description": None
    }]

    answer_entities = []
    # Format answer (as a time value, not an entity)

    answer_entities.append({
            "name": question[0][1],
            "mention": question[0][1],
            "span": None,
            "description": None
        })

    # Compose final output
    final = {
        "question": question[0][0],
        "question_entities": question_entities,
        "answer_entities": answer_entities,
        "lang": "en",
        "translations": None,
        "meta_data": {
            "id": None,
            "complexity_type": None,
            "category": None
        }
    }

    lines.append(final)


import ujson
from pathlib import Path

def write_jsonl(file_path, lines):
    """Create a .jsonl file and dump contents.
    file_path (unicode / Path): The path to the output file.
    lines (list): The JSON-serializable contents of each line.
    """
    data = [ujson.dumps(line, escape_forward_slashes=False) for line in lines]
    Path(file_path).open('w', encoding='utf-8').write('\n'.join(data))

write_jsonl("updated_durations_test.jsonl", lines)

lines
```
